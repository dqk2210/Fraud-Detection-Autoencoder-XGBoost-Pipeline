# 01 — Data Exploration

**Objective:** Load the raw BankSim dataset and perform initial exploration.

- Dataset shape, dtypes, memory usage
- Null / missing value check
- Class distribution (fraud vs normal)
- Basic descriptive statistics

```python
import sys, os
import pandas as pd
import numpy as np

# Ensure project root is importable
PROJECT_ROOT = os.path.abspath(os.path.join(os.getcwd(), '..'))
sys.path.insert(0, PROJECT_ROOT)

from src.config import get_paths, get_model_config

paths = get_paths()
cfg = get_model_config()['preprocessing']
print(f'Project root: {PROJECT_ROOT}')
```

## 1. Load Raw Data

```python
raw_csv = paths['data']['raw_csv']
df = pd.read_csv(raw_csv)

# Strip stray quote characters
for col in df.select_dtypes(include='object').columns:
    df[col] = df[col].str.strip("'\" ")

print(f'Loaded: {raw_csv}')
print(f'Shape: {df.shape[0]:,} rows × {df.shape[1]} columns')
df.head()
```

## 2. Data Types & Memory Usage

```python
print('--- Data Types ---')
print(df.dtypes)
print(f'\nMemory usage: {df.memory_usage(deep=True).sum() / 1e6:.1f} MB')
df.info()
```

## 3. Null / Missing Values

```python
null_counts = df.isnull().sum()
null_pct = (df.isnull().sum() / len(df) * 100).round(2)

null_report = pd.DataFrame({'Null Count': null_counts, 'Null %': null_pct})
print('--- Null Report ---')
print(null_report)
print(f'\nTotal nulls: {null_counts.sum()}')
```

## 4. Class Distribution

```python
target_col = cfg['target_col']

print('--- Class Distribution ---')
print(df[target_col].value_counts())
print(f'\nFraud rate: {df[target_col].mean():.4%}')
print(f'Imbalance ratio: 1:{int((1 - df[target_col].mean()) / df[target_col].mean())}')
```

## 5. Descriptive Statistics

```python
df.describe(include='all').T
```

## 6. Unique Values per Column

```python
for col in df.columns:
    print(f'{col:<20s} | nunique={df[col].nunique():>8,} | dtype={df[col].dtype}')
```
